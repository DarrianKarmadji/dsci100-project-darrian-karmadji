#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
summary_table <- tidy_womens_data |>
summarise(across(where(is.numeric), 
list(mean = ~mean(.x, na.rm = TRUE),
min = ~min(.x, na.rm = TRUE),
max = ~max(.x, na.rm = TRUE))))
summary_table
```

```r
missing_summary <- tidy_womens_data |>
summarise(across(everything(), ~sum(is.na(.x))))

missing_summary
```

Visualization 1: This first visualization basically gives a simplistic pliot that shows each team's win percentage in their conferences and in their regular seasons. Thus, we can use this, along with the classification of past tournament finishes for these teams in order to build a KNN classification system based off of the two different win percentages. 

```r
plot_rel1 <- ggplot(tidy_womens_data, aes(x = reg_wins_pct, y = conf_wins_pct)) +
geom_point(alpha = 0.5, color = "steelblue") +
labs(
title = "Regular Season vs. Conference Win Percentage",
x = "Regular Season Win Percentage",
y = "Conference Win Percentage")

plot_rel1
```

```r
plot_rel2 <- ggplot(tidy_womens_data, aes(x = tourney_finish, y = reg_wins_pct)) +
geom_boxplot(fill = "lightblue", color = "darkblue") +
labs(
title = "Regular Season Win Percentage by Tournament Finish",
x = "Tournament Finish",
y = "Regular Season Win Percentage") +
theme(axis.text.x = element_text(angle = 45, hjust = 1))

plot_rel2
```
